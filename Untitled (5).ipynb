```python
target_col = "Absenteeism time in hours"

print("Missing values before treatment:")
display(df.isnull().sum().sort_values(ascending=False))

# The target cannot be imputed meaningfully for supervised learning
df = df.dropna(subset=[target_col]).copy()

# Fill missing numerical values with each column's median
numeric_columns = df.select_dtypes(include="number").columns
for column in numeric_columns:
    if column != target_col:
        df[column] = df[column].fillna(df[column].median())

print("\nMissing values after treatment:")
display(df.isnull().sum().sort_values(ascending=False))
```

```python
def cap_outliers_iqr(dataframe, exclude_columns=None):
    dataframe = dataframe.copy()

    if exclude_columns is None:
        exclude_columns = []

    numeric_columns = dataframe.select_dtypes(include="number").columns

    for column in numeric_columns:
        if column in exclude_columns:
            continue

        q1 = dataframe[column].quantile(0.25)
        q3 = dataframe[column].quantile(0.75)
        iqr = q3 - q1

        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr

        dataframe[column] = dataframe[column].clip(
            lower=lower_bound,
            upper=upper_bound
        )

    return dataframe


df = cap_outliers_iqr(
    df,
    exclude_columns=["ID", "Absenteeism time in hours"]
)

print("Outlier capping completed.")
```

```python
import numpy as np
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestRegressor
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.tree import DecisionTreeRegressor

categorical_columns = [
    "Reason for absence",
    "Month of absence",
    "Day of the week",
    "Seasons",
    "Disciplinary failure",
    "Education",
    "Son",
    "Social drinker",
    "Social smoker",
    "Pet",
]

X_train_prep = X_train.copy()
X_test_prep = X_test.copy()

for col in X_train_prep.columns:
  if col not in categorical_columns and X_train_prep[col].dtype == "object":
    X_train_prep[col] = pd.to_numeric(
        X_train_prep[col]
        .astype(str)
        .str.replace(",", ".", regex=False)
        .str.strip(),
        errors="coerce",
    )
    X_test_prep[col] = pd.to_numeric(
        X_test_prep[col]
        .astype(str)
        .str.replace(",", ".", regex=False)
        .str.strip(),
        errors="coerce",
    )

numeric_features = [
    col
    for col in X_train_prep.select_dtypes(include="number").columns
    if col not in categorical_columns
]
categorical_features = [
    col for col in categorical_columns if col in X_train_prep.columns
]

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            Pipeline([
                ("imputer", SimpleImputer(strategy="median")),
                ("scaler", StandardScaler()),
            ]),
            numeric_features,
        ),
        (
            "cat",
            Pipeline([
                ("imputer", SimpleImputer(strategy="most_frequent")),
                ("onehot", OneHotEncoder(handle_unknown="ignore")),
            ]),
            categorical_features,
        ),
    ],
    remainder="drop",
)

X_train_scaled = preprocessor.fit_transform(X_train_prep)
X_test_scaled = preprocessor.transform(X_test_prep)

# Selecting regression algorithms
models = {
    "Linear Regression": LinearRegression(),
    "Ridge Regression": Ridge(random_state=42),
    "Decision Tree": DecisionTreeRegressor(random_state=42),
    "Random Forest": RandomForestRegressor(n_estimators=100, random_state=42),
}

# Store and evaluate regression results
scores = []
for name, model in models.items():
  model.fit(X_train_scaled, y_train)
  y_pred = model.predict(X_test_scaled)

  mae = mean_absolute_error(y_test, y_pred)
  rmse = np.sqrt(mean_squared_error(y_test, y_pred))
  r2 = r2_score(y_test, y_pred)

  scores.append({
      "Model": name,
      "MAE": round(mae, 2),
      "RMSE": round(rmse, 2),
      "R2": round(r2, 3),
  })

results_df = pd.DataFrame(scores)
print(results_df)
```
